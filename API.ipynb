```python
import pandas as pd
def weather_forecast():
    api_key="4f1384b8571d78afdaf9dda4e3ca39d2"
    city= input("Enter city name: ")
    request_url=f"http://api.openweathermap.org/geo/1.0/direct?q={city}&limit=5&appid={api_key}"
    response=requests.get(request_url)
    coordinates = response.json()
    lat=coordinates[0]['lat']
    lon=coordinates[0]['lon']
    
    request_url1=f"https://api.openweathermap.org/data/2.5/forecast?lat={lat}&lon={lon}&appid={api_key}"
    response1= requests.get(request_url1)
    weather_data= response1.json()
    # print(json.dumps(weather_data,indent=5))
    des=[]
    temp=[]
    pre=[]
    hum=[]
    ws=[]
    dt=[]
    v=[]
    for i in range(40):
        temp.append(weather_data['list'][i]['main']['temp'])
        pre.append(weather_data['list'][i]['main']['pressure'])
        hum.append(weather_data['list'][i]['main']['humidity'])
        des.append(weather_data['list'][i]['weather'][0]['description'])
        ws.append(weather_data['list'][i]['wind']['speed'])
        v.append(weather_data['list'][i]['visibility'])
        dt.append(weather_data['list'][i]['dt_txt'])
    df = pd.DataFrame({'Temp':temp,'Description':des,'Humidity':hum,'Pressure':pre,'Wind Speed':ws,
                           'Date-time':dt,'Visibility':v})
    print(df)
weather_forecast()
```

## Pagination

```python
import requests
url="https://jsonplaceholder.typicode.com/posts"
params={
    '_page':10,
    '_limit':5
}
response=requests.get(url,params=params)
data=response.json()
# print(json.dumps(data,indent=5))

for post in data:
    print(post['id'],"---",post['title'])
```

```python
import requests
url="https://jsonplaceholder.typicode.com/posts"
page=1
while page<=3:  
    params={
        '_page':page,'_limit':20
    }
    response=requests.get(url,params=params)
    data=response.json()
    for post in data:
        print(post['id'],"---",post['title'])
    page+=1
```
